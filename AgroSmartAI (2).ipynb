```python
df=df.drop_duplicates()
```

```python
#FOR categorial Data
categorial_cols=["State","Soil_Type","Season"]

for col in categorial_cols:
    df[col]=df[col].fillna(df[col].mode()[0])
    
```

```python
numerical_col=["Nitrogen","Phosphorus","Potassium","Temperature_C","Humidity_pct","Soil_pH","Rainfall_mm","Yield_tons_per_hectare"]

for col in numerical_col:
    df[col]=df[col].fillna(df[col].mean())
```

```python
numerical_cols = [
    "Nitrogen",
    "Phosphorus",
    "Potassium",
    "Temperature_C",
    "Humidity_pct",
    "Soil_pH",
    "Rainfall_mm"
]

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(col, "=", len(outliers), "outliers")
```

```python
x = df[
    [
        "Nitrogen",
        "Phosphorus",
        "Potassium",
        "Temperature_C",
        "Humidity_pct",
        "Soil_pH",
        "Rainfall_mm"
    ]
]

y = df["Crop"]
```

```python
print("Input Features Shape:", x.shape)
print("Target Shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.tree  import DecisionTreeClassifier
```

```python
dt=DecisionTreeClassifier(criterion='gini')
dt.fit(x_train,y_train)
print("Decision Tree Model Trained Successfully!")
```

```python
y_pred_dt = dt.predict(x_test)

print(y_pred_dt[:10])
```

```python
from sklearn.metrics import accuracy_score

dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy:", dt_accuracy * 100, "%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_dt))
```

```python
print(df["Crop"].value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(x_train, y_train)

print("Random Forest Model Trained Successfully!")
```

```python
y_pred_rf = rf_model.predict(x_test)

print(y_pred_rf[:10])
```

```python
from sklearn.metrics import accuracy_score

rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Decision Tree Accuracy:", round(dt_accuracy * 100, 2), "%")
print("Random Forest Accuracy:", round(rf_accuracy * 100, 2), "%")
```

```python
print("Decision Tree Accuracy:", round(dt.score(x_test, y_test) * 100, 2), "%")
print("Random Forest Accuracy:", round(rf_model.score(x_test, y_test) * 100, 2), "%")
```

```python
rf_model_2 = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    random_state=42
)

rf_model_2.fit(x_train, y_train)

rf_accuracy_2 = rf_model_2.score(x_test, y_test) * 100

print("Original Random Forest:", round(rf_accuracy * 100, 2), "%")
print("Tuned Random Forest:", round(rf_accuracy_2, 2), "%")
```

```python
import pandas as pd

feature_importance = pd.DataFrame({
    "Feature": x.columns,
    "Importance": rf_model_2.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_rf)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=rf_model.classes_
)

fig, ax = plt.subplots(figsize=(12, 12))
disp.plot(ax=ax, xticks_rotation=45, colorbar=False)
plt.title("Random Forest Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
y_pred_rf2 = rf_model_2.predict(x_test)

print("Tuned Random Forest Accuracy:",
      round(rf_model_2.score(x_test, y_test) * 100, 2), "%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_rf2))
```

```python
from sklearn.metrics import accuracy_score

y_pred_final = rf_model_2.predict(x_test)

final_accuracy = accuracy_score(y_test, y_pred_final)

print("Final Model: Random Forest")
print("Final Accuracy:", round(final_accuracy * 100, 2), "%")
```
